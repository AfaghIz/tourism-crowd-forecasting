# 27 Part B Random Forest\n
\n
Random Forest baseline for the Part B proxy-imputation task.\n
\n
Part B models POI attractiveness proxy, not true crowdedness.\n
The target is `log1p(wiki_pageviews_total)` on POIs with usable direct wiki signal.\n
Observed wiki scores are preserved; the model is only used to impute missing attractiveness scores.\n
This pass adds `is_tomb`, `is_fountain`, `is_gate`, and `has_valid_wiki_identity_but_missing_pageviews` to the feature set.\n

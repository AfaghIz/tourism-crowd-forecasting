# 27 Part B Random Forest\n
\n
Random Forest baseline for the Part B proxy-imputation task.\n
\n
Part B models POI attractiveness proxy, not true crowdedness.\n
The target is `log1p(wiki_pageviews_total)` on POIs with usable direct wiki signal.\n
Observed wiki scores are preserved; the model is only used to impute missing attractiveness scores.\n
This pass includes the subtype flags, the `has_valid_wiki_identity_but_missing_pageviews` indicator, and the new `family_*` semantic features.\n

```python
import json\n
import pandas as pd\n
import numpy as np\n
pd.set_option("display.max_columns", None)\n
pd.set_option("display.width", 180)\n
INPUT_PATH = "../data/processed/otm_pois_model_ready_with_nlp_features_exclusions.csv"\n
MODEL_TABLE_PATH = "../data/processed/otm_part_b_modeling_table.csv"\n
OUTPUT_PATH = "../data/processed/otm_part_b_rf_scores.csv"\n
METRICS_PATH = "../data/processed/otm_part_b_rf_metrics.json"\n
IMPORTANCES_PATH = "../data/processed/otm_part_b_rf_feature_importances.csv"\n
```

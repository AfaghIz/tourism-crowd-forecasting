# 24 Apply Wiki Review

Apply manual Wikipedia review decisions to the model-ready OTM POI file.

This notebook:
- reads the suspicious-match review sheet
- applies `keep`, `fix`, and `drop_wiki` decisions
- writes a reviewed POI dataset without overwriting the original
- creates a small re-enrichment queue for rows whose wiki title was manually fixed


```python
from pathlib import Path
import csv
import pandas as pd

root = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
review_path = root / 'data/processed/otm_wiki_match_review_completed.csv'
poi_path = root / 'data/processed/otm_pois_model_ready.csv'
reviewed_poi_path = root / 'data/processed/otm_pois_model_ready_wiki_reviewed.csv'
reenrich_path = root / 'data/processed/otm_wiki_reenrichment_queue.csv'

review_df = pd.read_csv(review_path)
poi_df = pd.read_csv(poi_path)
review_df[['poi_id','display_name_en','review_decision','corrected_wiki_title','notes']].head()
```

```python
reviewed_df = pd.read_csv(reviewed_poi_path)
reenrich_df = pd.read_csv(reenrich_path)
print('Reviewed rows:', len(reviewed_df))
print('Decision counts:')
print(reviewed_df['wiki_review_decision'].replace('', 'unreviewed').value_counts(dropna=False))
print('\nRows queued for re-enrichment:')
print(reenrich_df)
```

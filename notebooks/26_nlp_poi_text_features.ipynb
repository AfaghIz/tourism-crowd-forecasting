# 26 NLP POI Text Features

Lightweight rule-based NLP preprocessing for POI text metadata.

This notebook documents three small but useful text-processing tasks:
- semantic subtype feature extraction from POI text metadata
- broader tourism-experience family assignment derived from those subtype features
- potential duplicate POI detection using normalized names and spatial proximity

New subtype additions in this pass:
- `is_tomb`
- `is_fountain`
- `is_gate`


```python
from pathlib import Path
import pandas as pd
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 180)
ROOT = Path("..").resolve()
INPUT_PATH = "../data/processed/otm_pois_model_ready_wiki_reviewed.csv"
OUTPUT_PATH = "../data/processed/otm_pois_model_ready_with_nlp_features.csv"
DUP_OUTPUT_PATH = "../data/processed/otm_poi_duplicate_candidates.csv"
FAMILY_SCRIPT_PATH = ROOT / "scripts" / "generate_poi_families.py"
```
